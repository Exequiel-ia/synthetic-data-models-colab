# Ejecutar el laboratorio de datos sintéticos
Seleccione **Entorno de ejecución > Cambiar tipo de entorno de ejecución > T4 GPU** y ejecute las celdas en orden.

```python
REPO_URL = "https://github.com/Exequiel-ia/synthetic-data-models-colab.git"
PROJECT_DIR = "/content/synthetic-data-models-colab"
```

## 1. Descargar la versión más reciente desde GitHub

```python
import os, shutil, subprocess
os.chdir('/content')
if os.path.exists(PROJECT_DIR):
    shutil.rmtree(PROJECT_DIR)
subprocess.run(['git', 'clone', REPO_URL, PROJECT_DIR], check=True)
os.chdir(PROJECT_DIR)
print('Proyecto descargado en:', PROJECT_DIR)
```

## 3. Conectar Google Drive para conservar resultados

```python
from google.colab import drive
from pathlib import Path
drive.mount('/content/drive')
results = Path('/content/drive/MyDrive/Synthetic_Data_Lab/runs')
results.mkdir(parents=True, exist_ok=True)
os.environ['SYNTHETIC_RUNS_DIR'] = str(results)
print('Resultados persistentes en:', results)
```
